```python
import pandas as pd
```

```python
df = pd.read_csv("data/taptap_sample.csv")
print(df["score"].value_counts())  # 评分分布
print(df["content"].str.len().describe())  # 评论长度
```

```python
df = pd.read_csv("data/taptap_reviews_full.csv")

print(f"总数: {len(df)}")
print(f"去重后: {df['review_id'].nunique()}")
print(f"\n字段缺失率:")
print(df.isnull().mean().sort_values(ascending=False))

print(f"\n时间跨度: {df['created_date'].min()} → {df['created_date'].max()}")
print(f"覆盖天数: {pd.to_datetime(df['created_date']).nunique()}")
```

```python
print(df["score"].value_counts().sort_index())
print(f"\n均分: {df['score'].mean():.2f}")
print(f"1-2星占比: {(df['score']<=2).mean():.1%}")
print(f"4-5星占比: {(df['score']>=4).mean():.1%}")
```
